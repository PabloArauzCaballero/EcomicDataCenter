# Verificación del plan de 300 personalidades

Inspecciona los archivos locales preparados el 8 de octubre de 2026. No ejecuta recolección ni modifica datos. Las 108.000 filas son tareas pendientes, no posts recuperados.

```python
from pathlib import Path
import json, csv, gzip, hashlib, collections
base = Path.cwd()
candidates = [base, base / 'docs/research/personalidades-300-cinco-anos', base / 'EcomicDataCenter/docs/research/personalidades-300-cinco-anos']
folder = next(p for p in candidates if (p / 'baseline.json').exists())
baseline = json.loads((folder / 'baseline.json').read_text(encoding='utf-8'))
print(json.dumps(baseline, indent=2, ensure_ascii=False))
```

```python
validation = json.loads((folder / 'validation.json').read_text(encoding='utf-8'))
for artifact in validation['outputs']:
    actual = hashlib.sha256((folder / artifact['file']).read_bytes()).hexdigest()
    assert actual == artifact['sha256'], artifact['file']
with gzip.open(folder / 'coverage-initial.csv.gz', 'rt', encoding='utf-8-sig', newline='') as stream:
    rows = list(csv.DictReader(stream))
assert len(rows) == len({r['task_id'] for r in rows}) == 108000
assert all(r['posts_discovered'] == r['posts_captured'] == r['posts_reviewed'] == '' for r in rows)
assert all(r['enumeration_status'] == 'NOT_STARTED' for r in rows)
print('PASS: huellas, claves únicas, conteos desconocidos y 108.000 tareas sin iniciar')
print(collections.Counter(r['platform'] for r in rows))
```

La validación acredita la coherencia de la planificación. No acredita identidad de cuentas, disponibilidad de acceso, cobertura del historial ni calidad de anotaciones todavía no realizadas.
